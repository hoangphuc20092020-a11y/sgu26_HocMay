## Bài 1. Tạo Tensor cơ bản
### Tạo một Tensor số nguyên kích thước 2 × 3 và hiển thị kết quả.

```python
import torch
a =torch.tensor(
    [[1,2,3],
     [4,5,6]],
    dtype = torch.int32
)
print(a)
```

#### Yêu cầu: Cho biết kích thước, số chiều và kiểu dữ liệu của Tensor a.

- kích thước:(2,3) 
- số chiều: 2 chiều
- kiểu dữ liệu: torch.int32

## Bài 2. Tạo các Tensor thường dùng
### Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

```python
import torch
a = torch.rand(3,4)
b = torch.randn(3,4)
c = torch.zeros(2,3)
d = torch.ones(2,3)
e = torch.eye(4)

print(a)
print(b)
print(c)
print(d)
print(e)
```

#### Yêu cầu:
- Nhận xét sự khác nhau giữa torch.rand() và torch.randn().
- Tạo thêm một Tensor 5 × 5 chứa toàn giá trị 10.

- sự khác nhau torch.rand()tạo bảng chứ số [0,1] còn torch.randn() với số ngâu nhiên quanh 0 có âm và dương

```python
import torch
f = torch.full((5,5), 10)
print(f)
```

## Bài 3. Kiểm tra thông tin Tensor
### Sử dụng shape, size(), ndim và dtype để kiểm tra các thuộc tính của Tensor.

```python
import torch
a = torch.zeros(2, 3, 4)
print("Tensor:")
print(a)
print("shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

#### yêu cầu:Yêu cầu: Giải thích ý nghĩa của torch.Size([2, 3, 4]).
- torch.Size([2, 3, 4]) cho biết tersor a có 3 chiều và kích thước lần lượt 2x3x4 nghĩa tersor có 2 khối, 3 hàng, 4 cột

## Bài 4. Slicing và Indexing
### Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

```python
import torch
a = torch.randn(3, 4, 5)
print("original tensor:")
print(a)
print("tensor a[1]:")
print(a[1])
print("tensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

### Yêu cầu:
- Cho biết shape của a[1] và a[1:, 2:4].
- Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

- a[1].shape = [4,5], a[1:, 2:4] = [2,2,5]
- note:
        a[1]       # Indexing
        a[1:3]     # Slicing
        a[-2:]     # Slicing

```python
print(a[:,:,-2:])
```

## Bài 5. Reshape và Flatten
### Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

```python
import torch
a = torch.randn(3, 4)
print("original:")
print(a)
print("\nFlatten: ")
print(a.ravel())
print("\nReshape to 3 x 2 x 2: ")
print(a.reshape(3,2,2))
```

#### Yêu cầu:
- Kiểm tra số phần tử trước và sau khi reshape.
- Thử reshape Tensor thành kích thước 2 × 6.

#### Trả lời
-Số phần tử trước và sau khi reshape vẫn giống nhau là 12 phần tử. Reshape chỉ thay đổi shape/cách tổ chức dữ liệu, không làm thay đổi số lượng phần tử.

```python
print("reshape to 2 x 6")
print(a.reshape(2,6))
```

## Bài 6. Phép toán và nhân ma trận
### Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

```python
import torch
a = torch.randn(2,3)
b = torch.randn(2,3)
print("a:")
print(a)
print("b:")
print(b)
print("\n a+b")
print(a+b)
print("\n a*b")
print(a*b)
print("\n a/b")
print(a/b)
print("\n a^2")
print(a**2)
print("\nMatrix multiplication: ")
print(a @ b.T)
```
