## Bài 7. Các hàm thống kê trên Tensor
### Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

```python
import torch
a = torch.randn(3,4)
print("a")
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStanderd devitation")
print(torch.std(a, dim = 0))
print("\nCumulative sum: ")
print(torch.cumsum(a, dim =0))
```

## Bài 8. Autograd cơ bản
### Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
import torch
x = torch.tensor(3.6,requires_grad= True)
y = x*x
y.backward()
print("x=", x)
print("y=", y)
print("x.grad=", x.grad)
```

#### Yêu cầu:
- Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.
- Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

#### trả lời
- tính thủ công với y = x^2 đạo hàm = 7.2 bằng x.grad 

```python
import torch
x = torch.tensor(3.6,requires_grad= True)
y = 3*x**2 + 2*x + 1
y.backward()
print("x=", x)
print("y=", y)
print("x.grad=", x.grad)
```

## Bài 9. Gradient Descent tìm hệ số đa thức
### Sinh dữ liệu từ đa thức y = x^2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

polynomial = np.poly1d([1,2,3]) # đa thứ thật y = x^2 + 2x + 3
N =20
X =np.random.randn(N, 1) * 5 # xin ngẫu nhiên 20 dữ liệu X 
Y = polynomial(X) # kết qua y qua tường X
XX = np.hstack( # np.hstack ghép theo chiều ngang
    [
        X*X,X,np.ones_like(X)
    ]
)
w = torch.randn(3, 1, requires_grad = True)
x = torch.tensor(XX,dtype=torch.float32)
y = torch.tensor(Y,dtype = torch.float32)

optimizer =torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:") # thấy hệ số ngẫu nhiên điểm xuất phát
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w # nhân 2 ma trận kết hợp x a b c lại cho ra dạng[ [a*x1² + b*x1 + c],[a*x2² + b*x2 + c],[a*x3² + b*x3 + c],...]
    mse = torch.mean(torch.square(y-y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

#### Yêu cầu:
- So sánh ba hệ số học được với [1, 2, 3].
- Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().
#### trả lời
- hệ số học được là [1,0108,1.9685,2.3661] so với [1, 2, 3] lệch tới [0.0094,0.0020,1.0360]
- optimizer.zero_grad(): xóa đi gradient cũ để chuẩn bị tính gradient mới,xóa để tránh việc cộng dồn
-  mse.backward():Autograd tìm xem nên sửa a, b, c thế nào để sai số giảm xuống thật thấp
- optimizer.step(): khi biết được thông tin sửa thì optimizer.step() như người update lại a b c

## Bài 10. Giải hệ phương trình bằng Autograd
### Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 5000
optimizer = torch.optim.NAdam([A, B, C, D], lr = 0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A=",A)
print("B",B)
print("C",C)
print("D", D)
```

#### Yêu cầu:
- Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.
- Nhận xét vì sao bài toán có thể có nhiều nghiệm.
